```python
# fast: no enrichment, just the exchange/asset_class/instrument filter
df_uk_equity_raw = justetf_scraping.load_overview(
    exchange="XLON",
    asset_class="class-equity",
    instrument="ETF",
    local_country="GB",
    currency="GBP",
    enrich=False,
)
print(f"UK-listed equity ETFs (unenriched): {len(df_uk_equity_raw)}")
df_uk_equity_raw[["ticker", "name", "inception_date", "age_in_years", "currency", "ter", "size"]].head(10)
```

```python
tickers = df_uk_equity_raw["ticker"].dropna().unique().tolist()
print(f"Unique tickers to test: {len(tickers)}")

import yfinance as yf

full_tickers = [t + ".L" for t in tickers]

# batch download — much faster than looping per ticker
data = yf.download(full_tickers, period="5d", group_by="ticker",
                    auto_adjust=True, progress=True, threads=True)

print("Download shape:", data.shape)
```

```python
failed = ['FWDS.L','OCTF.L','QUGB.L','USCO.L','PAGB.L','PQWP.L',
          'MWDP.L','QRMW.L','XNZG.L','XEQ6.L','SX5D.L','XEQ8.L','CECL.L',
          'EMRG.L','XEQ2.L','WMSE.L','CSUKX.L','SUUE.L','CC1G.L','CSUKS.L',
          'TURL.L','GQWX.L','S3RA.L','USAC.L','SGQP.L','XEQ4.L','EMUE.L',
          'IM12.L','WSRI.L','XDND.L','XBAS.L','WMMV.L','WMMS.L','WMFS.L',
          'CN1.L','CHSI.L','CWEG.L','XDGM.L','MVEE.L','JPNY.L']

good_tickers = [t for t in full_tickers if t not in failed]
print(f"Tickers to pull full history for: {len(good_tickers)}")
```

```python
import pandas as pd, yfinance as yf, time
from pathlib import Path

Path("data/processed").mkdir(parents=True, exist_ok=True)

CHUNK_SIZE = 50
all_prices = []
currency_map = {}
failed_tickers = []

for i in range(0, len(good_tickers), CHUNK_SIZE):
    chunk = good_tickers[i:i+CHUNK_SIZE]
    print(f"Batch {i//CHUNK_SIZE + 1}/{(len(good_tickers)-1)//CHUNK_SIZE + 1}  "
          f"({i}-{min(i+CHUNK_SIZE, len(good_tickers))} of {len(good_tickers)})")

    try:
        data = yf.download(chunk, period="max", group_by="ticker",
                            auto_adjust=True, progress=False, threads=True)
    except Exception as e:
        print(f"  batch error: {e}")
        failed_tickers.extend(chunk)
        continue

    for t in chunk:
        try:
            df_t = data[t].dropna(how="all")
            if df_t.empty:
                failed_tickers.append(t)
                continue
            df_t = df_t.reset_index()
            df_t["ticker"] = t.replace(".L", "")
            all_prices.append(df_t[["ticker","Date","Open","High","Low","Close","Volume"]])

            # currency check — needed for the GBX/pence fix
            try:
                currency_map[t.replace(".L","")] = yf.Ticker(t).fast_info.currency
            except Exception:
                currency_map[t.replace(".L","")] = "unknown"
        except Exception:
            failed_tickers.append(t)

    # periodic safety save every 10 batches (~500 tickers)
    if (i // CHUNK_SIZE) % 10 == 0 and all_prices:
        pd.concat(all_prices, ignore_index=True).to_csv(
            "data/processed/_prices_partial.csv", index=False)
        print(f"  [checkpoint saved: {len(all_prices)} tickers so far]")

print(f"\nDone. Succeeded: {len(all_prices)}  Failed: {len(failed_tickers)}")
```

```python
# 1. Consolidate all price data collected so far
prices_df = pd.concat(all_prices, ignore_index=True)
prices_df.columns = ["ticker","date","open","high","low","close","volume"]
prices_df["date"] = pd.to_datetime(prices_df["date"])

actual_tickers_with_data = prices_df["ticker"].unique()
print(f"Tickers with real price data: {len(actual_tickers_with_data)}")
print(f"Total price rows: {len(prices_df):,}")
print(f"Date range: {prices_df['date'].min().date()} to {prices_df['date'].max().date()}")

# true failure list = requested minus what actually landed in prices_df
requested = [t.replace(".L","") for t in good_tickers]
truly_failed = sorted(set(requested) - set(actual_tickers_with_data))
print(f"Truly missing: {len(truly_failed)}")
```

```python
# 2. Build metadata table: merge justETF fields + currency/scale fix
meta = df_uk_equity_raw.reset_index().rename(columns={"isin": "isin"})
meta["ticker_clean"] = meta["ticker"]
meta = meta[meta["ticker_clean"].isin(actual_tickers_with_data)].copy()

meta["currency_actual"] = meta["ticker_clean"].map(currency_map)
meta["price_scale"] = meta["currency_actual"].map({"GBp": 100, "GBP": 1, "USD": 1}).fillna(1)

meta_final = meta[["ticker_clean","isin","name","inception_date","age_in_years",
                    "currency","currency_actual","price_scale","ter","size"]]
meta_final = meta_final.rename(columns={"ticker_clean": "ticker",
                                         "currency": "base_currency",
                                         "currency_actual": "quote_currency"})
print(f"Metadata rows: {len(meta_final)}")
meta_final.head(10)
```

```python
# 3. Write both CSVs
prices_df.sort_values(["ticker","date"]).to_csv(
    "data/processed/uk_etf_prices.csv", index=False)
meta_final.to_csv("data/processed/uk_etf_metadata.csv", index=False)

import os
print("prices.csv size:", round(os.path.getsize("data/processed/uk_etf_prices.csv")/1e6, 1), "MB")
print("metadata.csv size:", round(os.path.getsize("data/processed/uk_etf_metadata.csv")/1e6, 2), "MB")
```

```python
# 1. Dedupe metadata — keep the first occurrence per ticker
dupe_count = meta_final["ticker"].duplicated().sum()
print(f"Duplicate ticker rows found: {dupe_count}")
meta_final = meta_final.drop_duplicates(subset="ticker", keep="first").reset_index(drop=True)
print(f"Metadata rows after dedupe: {len(meta_final)}")

# 2. Normalize prices to GBP — apply price_scale (100 for GBX/pence, 1 for GBP/USD)
scale_map = meta_final.set_index("ticker")["price_scale"].to_dict()
prices_df["scale"] = prices_df["ticker"].map(scale_map).fillna(1)

for col in ["open", "high", "low", "close"]:
    prices_df[col] = prices_df[col] / prices_df["scale"]

prices_df = prices_df.drop(columns="scale")

# sanity check: no more 100x jumps between similar funds
print(prices_df.groupby("ticker")["close"].last().describe())
```

```python
# 3. Re-save final, corrected files
prices_df.sort_values(["ticker","date"]).to_csv(
    "data/processed/uk_etf_prices.csv", index=False)
meta_final.to_csv("data/processed/uk_etf_metadata.csv", index=False)

import os
Path("data/processed/_prices_partial.csv").unlink(missing_ok=True)  # cleanup checkpoint
print("prices.csv:", round(os.path.getsize("data/processed/uk_etf_prices.csv")/1e6, 1), "MB")
print("metadata.csv:", round(os.path.getsize("data/processed/uk_etf_metadata.csv")/1e6, 2), "MB")
print("Done — downstream notebooks now just pd.read_csv() these two files.")
```

```python
unknown_currency = meta_final[meta_final["quote_currency"] == "unknown"]
print(f"Tickers with unresolved currency: {len(unknown_currency)}")
unknown_currency[["ticker","name","base_currency","quote_currency"]]
```

```python
# flag any suspiciously high closes — likely un-scaled GBX funds
latest_close = prices_df.groupby("ticker")["close"].last().sort_values(ascending=False)
print(latest_close.head(15))
```

```python
import time

retry_currency = {}
still_unknown = []

for t in unknown_currency["ticker"].tolist():
    try:
        c = yf.Ticker(t + ".L").fast_info.currency
        retry_currency[t] = c if c else "unknown"
    except Exception:
        still_unknown.append(t)
    time.sleep(0.3)  # small delay — likely rate-limiting caused the original failures

print(f"Resolved on retry: {sum(1 for v in retry_currency.values() if v != 'unknown')}")
print(f"Still unresolved: {len(still_unknown) + sum(1 for v in retry_currency.values() if v == 'unknown')}")
pd.Series(retry_currency).value_counts()
```

```python
# apply the newly-resolved currencies: rescale only rows that turned out GBp
new_scale = {t: (100 if c == "GBp" else 1) for t, c in retry_currency.items()}

affected = [t for t, s in new_scale.items() if s == 100]
print(f"Tickers needing rescale (GBX->GBP): {len(affected)}")

mask = prices_df["ticker"].isin(affected)
for col in ["open", "high", "low", "close"]:
    prices_df.loc[mask, col] = prices_df.loc[mask, col] / 100

# update metadata with resolved currencies
meta_final["quote_currency"] = meta_final.apply(
    lambda r: retry_currency.get(r["ticker"], r["quote_currency"]), axis=1)
meta_final["price_scale"] = meta_final.apply(
    lambda r: new_scale.get(r["ticker"], r["price_scale"]), axis=1)

# re-check the outlier closes
latest_close = prices_df.groupby("ticker")["close"].last().sort_values(ascending=False)
print(latest_close.head(15))
print(f"\nStill-unknown tickers remaining: {(meta_final['quote_currency']=='unknown').sum()}")
```

```python
unresolved_tickers = meta_final.loc[meta_final["quote_currency"]=="unknown", "ticker"].tolist()

last_close = prices_df[prices_df["ticker"].isin(unresolved_tickers)].groupby("ticker")["close"].last()

THRESHOLD = 1000  # GBP-quoted ETF units essentially never exceed this
inferred_scale = last_close.apply(lambda x: 100 if x > THRESHOLD else 1)

print(f"Inferred as GBX (pence) -> rescale: {(inferred_scale == 100).sum()}")
print(f"Inferred as already GBP: {(inferred_scale == 1).sum()}")
inferred_scale.sort_values(ascending=False).head(20)
```

```python
affected = inferred_scale[inferred_scale == 100].index.tolist()
mask = prices_df["ticker"].isin(affected)
for col in ["open","high","low","close"]:
    prices_df.loc[mask, col] = prices_df.loc[mask, col] / 100

meta_final["price_scale"] = meta_final.apply(
    lambda r: inferred_scale.get(r["ticker"], r["price_scale"]), axis=1)
meta_final["quote_currency"] = meta_final.apply(
    lambda r: ("GBp (inferred)" if inferred_scale.get(r["ticker"]) == 100
               else "GBP (inferred)") if r["ticker"] in unresolved_tickers
               else r["quote_currency"], axis=1)

# re-save final files
prices_df.sort_values(["ticker","date"]).to_csv("data/processed/uk_etf_prices.csv", index=False)
meta_final.to_csv("data/processed/uk_etf_metadata.csv", index=False)

latest_close_final = prices_df.groupby("ticker")["close"].last().sort_values(ascending=False)
print(latest_close_final.head(10))
print(f"\nMax close now: {latest_close_final.max():.2f}")
```
